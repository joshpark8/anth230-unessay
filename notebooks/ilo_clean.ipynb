```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('../data/dirty/ILO/EAR_4MTH_SEX_CUR_NB_A-20250408T0015.csv')
```

```python
subset_f = df.query(
    '`indicator.label` == "Average monthly earnings of employees by sex"'
    'and `sex.label` == "Sex: Female"'
)
subset_m = df.query(
    '`indicator.label` == "Average monthly earnings of employees by sex"'
    'and `sex.label` == "Sex: Male"'
)

subset_usd_f = df.query(
    '`indicator.label` == "Average monthly earnings of employees by sex"'
    'and `sex.label` == "Sex: Female"'
    'and `classif1.label` == "Currency: U.S. dollars"'
)
subset_usd_m = df.query(
    '`indicator.label` == "Average monthly earnings of employees by sex"'
    'and `sex.label` == "Sex: Male"'
    'and `classif1.label` == "Currency: U.S. dollars"'
)

subset_ppp_f = df.query(
    '`indicator.label` == "Average monthly earnings of employees by sex"'
    'and `sex.label` == "Sex: Female"'
    'and `classif1.label` == "Currency: 2021 PPP $"'
)
subset_ppp_m = df.query(
    '`indicator.label` == "Average monthly earnings of employees by sex"'
    'and `sex.label` == "Sex: Male"'
    'and `classif1.label` == "Currency: 2021 PPP $"'
)
```

```python
idx_f = subset_f.groupby("ref_area.label")["time"].idxmax()
idx_m = subset_m.groupby("ref_area.label")["time"].idxmax()
idx_usd_f = subset_usd_f.groupby("ref_area.label")["time"].idxmax()
idx_usd_m = subset_usd_m.groupby("ref_area.label")["time"].idxmax()
idx_ppp_f = subset_ppp_f.groupby("ref_area.label")["time"].idxmax()
idx_ppp_m = subset_ppp_m.groupby("ref_area.label")["time"].idxmax()

latest_by_country_f = df.loc[idx_f].reset_index(drop=True)
latest_by_country_m = df.loc[idx_m].reset_index(drop=True)
latest_by_country_usd_f = df.loc[idx_usd_f].reset_index(drop=True)
latest_by_country_usd_m = df.loc[idx_usd_m].reset_index(drop=True)
latest_by_country_ppp_f = df.loc[idx_ppp_f].reset_index(drop=True)
latest_by_country_ppp_m = df.loc[idx_ppp_m].reset_index(drop=True)
```

```python
print(f'f: {latest_by_country_f.shape}')
print(f'm: {latest_by_country_m.shape}')
print(f'f usd: {latest_by_country_usd_f.shape}')
print(f'm usd: {latest_by_country_usd_m.shape}')
print(f'f ppp: {latest_by_country_ppp_f.shape}')
print(f'm ppp: {latest_by_country_ppp_m.shape}')

print(latest_by_country_f.head())
print(latest_by_country_m.head())
print(latest_by_country_usd_f.head())
print(latest_by_country_usd_m.head())
print(latest_by_country_ppp_f.head())
print(latest_by_country_ppp_m.head())
```

```python
latest_by_country_f.to_csv("../data/clean/avg_mtly_f.csv", index=False)
latest_by_country_m.to_csv("../data/clean/avg_mtly_m.csv", index=False)

latest_by_country_usd_f.to_csv("../data/clean/avg_mtly_usd_f.csv", index=False)
latest_by_country_usd_m.to_csv("../data/clean/avg_mtly_usd_m.csv", index=False)

latest_by_country_ppp_f.to_csv("../data/clean/avg_mtly_ppp_f.csv", index=False)
latest_by_country_ppp_m.to_csv("../data/clean/avg_mtly_ppp_m.csv", index=False)
```
